### Concept of Multilevel and Multiple Inheritance.

##### Create following classes
###### Class A → method display()
###### Class B → method display()
###### Class C (inherits A, B)
###### Call display() and observe which method runs.
###### Print MRO using: print(ClassName.__mro__)

```python
class A:
    def display(self):
        print("Call Form CLass A....")

class B:
    def display(self):
        print("Call from Class B.....")

class C(A,B):
    pass

obj=C()
obj.display()
print(C.__mro__)
```

#####  Create following classes
###### Person
###### Employee (inherits Person)
###### Student (inherits Person)
###### WorkingStudent (inherits Employee & Student)
###### Show how data is accessed from all classes.

```python
class Person:
    def __init__(self, name):
        self.name = name
    def show_person(self):
        print("Name:", self.name)

class Employee(Person):
    def __init__(self, name, salary):
        Person.__init__(self, name)
        self.salary = salary

    def show_employee(self):
        print("Salary:", self.salary)

class Student(Person):
    def __init__(self, name, marks):
        Person.__init__(self, name)
        self.marks = marks

    def show_student(self):
        print("Marks:", self.marks)

class WorkingStudent(Employee, Student):
    def __init__(self, name, salary, marks):
        Employee.__init__(self, name, salary)
        Student.__init__(self, name, marks)

    def show_all(self):
        self.show_person()
        self.show_employee()
        self.show_student()
# Object creation
ws = WorkingStudent("Sayan", 25000, 85)
# Access all data
ws.show_all()
```
