```python
# Accept n integers
n = int(input("Enter number of elements: "))
nums = []

for i in range(n):
    nums.append(int(input("Enter number: ")))

# Even numbers
print("Even numbers:")
for num in nums:
    if num % 2 == 0:
        print(num, end=" ")

# Prime count
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

prime_count = sum(1 for num in nums if is_prime(num))
print("\nNumber of prime numbers:", prime_count)

# Second largest
nums = list(set(nums))  # remove duplicates
nums.sort()

if len(nums) >= 2:
    print("Second largest number:", nums[-2])
else:
    print("Not enough elements")
```

```python
s = input("Enter a string: ")

# Frequency of characters
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

print("Character frequencies:")
for k, v in freq.items():
    print(k, ":", v)

# Remove duplicates
result = ""
for ch in s:
    if ch not in result:
        result += ch
print("String after removing duplicates:", result)

# Palindrome check
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
n = int(input("Enter n: "))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
# Take matrix dimensions from user
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

# Take matrix input
matrix = []
print("Enter matrix:")
for i in range(n):
    while True:
        row = list(map(int, input(f"Row {i+1}: ").split()))
        if len(row) == m:
            matrix.append(row)
            break
        else:
            print(f"Please enter exactly {m} numbers.")

# Print original matrix
print("\nOriginal Matrix:")
for row in matrix:
    print(row)

# ---- TRANSPOSE ----
print("\nTranspose:")
for i in range(m):
    for j in range(n):
        print(matrix[j][i], end=" ")
    print()

# ---- ROW SUMS ----
print("\nRow Sums:")
for i in range(n):
    total = 0
    for j in range(m):
        total = total + matrix[i][j]
    print(f"  Row {i+1} sum = {total}")

# ---- COLUMN SUMS ----
print("\nColumn Sums:")
for i in range(m):
    total = 0
    for j in range(n):
        total = total + matrix[j][i]
    print(f"  Column {i+1} sum = {total}")

# ---- SYMMETRY CHECK ----
if n == m:
    symmetric = True                  # Assume symmetric at first
    for i in range(n):
        for j in range(n):
            if matrix[i][j] != matrix[j][i]:
                symmetric = False     # Found a mismatch, so not symmetric
    if symmetric:
        print("\nMatrix is Symmetric")
    else:
        print("\nMatrix is Not Symmetric")
else:
    print("\nNot a square matrix, so symmetry check is not applicable.")
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_grade(self):
        if self.marks >= 90:
            return 'A'
        elif self.marks >= 75:
            return 'B'
        elif self.marks >= 50:
            return 'C'
        else:
            return 'Fail'

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("Grade:", self.get_grade())
        print("-------------------")

# Create objects
s1 = Student("Alice", 92)
s2 = Student("Bob", 76)
s3 = Student("Charlie", 45)

# Display details
s1.display()
s2.display()
s3.display()
```
