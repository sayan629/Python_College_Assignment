## Inheritance

```python
# Q1 Create a class Vehicle with attributes: brand, speed and one method: show() to display details.
#     Create a class Car that inherits from Vehicle and adds attribute: fuel_type displays all details.


# Base class
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def show(self):
        print(f"Brand : {self.brand}")
        print(f"Speed : {self.speed} km/h")


# Derived class
class Car(Vehicle):
    def __init__(self, brand, speed, fuel_type):
        Vehicle.__init__(self, brand, speed)   # Call base class constructor
        self.fuel_type = fuel_type

    def show(self):
        Vehicle.show(self)                     # Call base class show()
        print(f"Fuel Type : {self.fuel_type}")


# Create objects
print("---- Car Details ----")
car1 = Car("Toyota", 180, "Petrol")
car1.show()

print()

car2 = Car("Tesla", 250, "Electric")
car2.show()
```

```python
# Q2 Create a class Employee with attributes: emp_id, salary. Create a derived class Manager that
# adds attribute: department. Use super() to initialize base class attributes


# Base class
class Employee:
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary

    def show(self):
        print(f"Employee ID : {self.emp_id}")
        print(f"Salary      : Rs.{self.salary}")


# Derived class
class Manager(Employee):
    def __init__(self, emp_id, salary, department):
        super().__init__(emp_id, salary)       # Initialize base class using super()
        self.department = department

    def show(self):
        super().show()                         # Call base class show()
        print(f"Department  : {self.department}")


# Create objects
print("---- Manager 1 ----")
m1 = Manager("E101", 75000, "Engineering")
m1.show()

print()

print("---- Manager 2 ----")
m2 = Manager("E102", 90000, "Human Resources")
m2.show()
```

```python
# Q3 Create a class Shape with: method area() that prints “Area not defined”. Create a class
# Rectangle that inherits Shape overrides area() to calculate area using length and breadth.


# Base class
class Shape:
    def area(self):
        print("Area not defined")       # Default message


# Derived class overrides area()
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):                     # Override base class method
        result = self.length * self.breadth
        print(f"Length  : {self.length}")
        print(f"Breadth : {self.breadth}")
        print(f"Area    : {result} sq. units")


# Base class object
print("---- Shape Object ----")
s = Shape()
s.area()

print()

# Derived class object
print("---- Rectangle Object ----")
r = Rectangle(10, 5)
r.area()

print()

print("---- Rectangle Object 2 ----")
r2 = Rectangle(7, 3)
r2.area()
```

```python
# Q4 Create a class Account with attributes: account_no, balance and method: display_balance().
# Create a class SavingsAccount that inherits Account adds interest rate calculates final balance after
# adding interest (interest is 6% of balance)


# Base class
class Account:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance    = balance

    def display_balance(self):
        print(f"Account No : {self.account_no}")
        print(f"Balance    : Rs.{self.balance}")


# Derived class
class SavingsAccount(Account):
    def __init__(self, account_no, balance):
        super().__init__(account_no, balance)
        self.interest_rate = 6             # Fixed interest rate of 6%

    def calculate_interest(self):
        interest      = (self.interest_rate / 100) * self.balance
        final_balance = self.balance + interest
        return interest, final_balance

    def display_balance(self):
        super().display_balance()          # Show base details
        interest, final_balance = self.calculate_interest()
        print(f"Interest Rate  : {self.interest_rate}%")
        print(f"Interest Earned: Rs.{interest}")
        print(f"Final Balance  : Rs.{final_balance}")


# Create objects
print("---- Savings Account 1 ----")
sa1 = SavingsAccount("SA1001", 50000)
sa1.display_balance()

print()

print("---- Savings Account 2 ----")
sa2 = SavingsAccount("SA1002", 120000)
sa2.display_balance()
```

```python
# Q5 Create a class Mobile with attributes: brand, price. Create a class Smartphone that inherits
# Mobile adds camera_megapixels and prints all details


# Base class
class Mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def show(self):
        print(f"Brand : {self.brand}")
        print(f"Price : Rs.{self.price}")


# Derived class
class Smartphone(Mobile):
    def __init__(self, brand, price, camera_megapixels):
        super().__init__(brand, price)
        self.camera_megapixels = camera_megapixels

    def show(self):
        super().show()                    # Display base class details
        print(f"Camera: {self.camera_megapixels} MP")


# Create objects
print("---- Smartphone 1 ----")
sp1 = Smartphone("Samsung", 25000, 108)
sp1.show()

print()

print("---- Smartphone 2 ----")
sp2 = Smartphone("Apple", 80000, 48)
sp2.show()

print()

print("---- Smartphone 3 ----")
sp3 = Smartphone("OnePlus", 35000, 64)
sp3.show()
```
