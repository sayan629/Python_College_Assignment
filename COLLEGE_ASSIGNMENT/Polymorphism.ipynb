### Polymorphism Concept

##### Q1 Write a Python program with a function add(a, b) that works for:
###### Two integers
###### Two strings

```python
def add(a, b):
    return a + b
print(add(10, 20))        # integers
print(add("Hello ", "World"))  # strings
```

##### Q2 Create a base class Shape with method area(). Create subclasses:
###### Rectangle → calculate area
###### Circle → calculate area
###### Use the same method name area() for different shapes.

```python
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w
        
    def area(self):
        return self.l * self.w

class Circle(Shape):
    def __init__(self, r):
        self.r = r
        
    def area(self):
        return 3.14 * self.r * self.r

r = Rectangle(5, 4)
c = Circle(3)

print("Rectangle Area:", r.area())
print("Circle Area:", c.area())
```

##### Q3 Create a Payment System:
###### Class Payment → method pay()
###### Subclasses:
###### CreditCard
###### UPI
###### Cash
###### Each should implement pay() differently.

```python
class Payment:
    def pay(self):
        print("Processing payment...")

class CreditCard(Payment):
    def pay(self):
        print("Paid using Credit Card")

class UPI(Payment):
    def pay(self):
        print("Paid using UPI")

class Cash(Payment):
    def pay(self):
        print("Paid using Cash")

p1 = CreditCard()
p2 = UPI()
p3 = Cash()

p1.pay()
p2.pay()
p3.pay()
```
