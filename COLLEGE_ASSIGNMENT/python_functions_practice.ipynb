# Python Program Using Function

```python
def square(n):
    result = n * n
    return result
num=int(input("Enter Number: "))
print("Square is:", square(num))
```

```python
def simple_interest(p, r, t):
    si = (p * r * t) / 100
    return si
print("Simple Interest is:", simple_interest(1000, 5, 2))
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact
print("Factorial is:", factorial(5))
```

```python
def check_prime(n):
    if n <= 1:
        return "Not Prime"
    for i in range(2, n):
        if n % i == 0:
            return "Not Prime" 
    return "Prime"
print(check_prime(7))
```

```python
def max_min(lst):
    maximum = lst[0]
    minimum = lst[0]
    for num in lst:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    return maximum, minimum

numbers = [10, 5, 30, 2, 8]
print("Max and Min:", max_min(numbers))
```

```python
def count_even(lst):
    count = 0
    for num in lst:
        if num % 2 == 0:
            count = count + 1
    return count

print("Even numbers:", count_even([1, 2, 3, 4, 6]))
```

```python
def tuple_length(t):
    count = 0
    for i in t:
        count = count + 1
    return count

print("Length is:", tuple_length((1, 2, 3, 4)))
```

```python
def print_values(d):
    for key in d:
        print(d[key])

student = {"name": "Sayan", "age": 22, "marks": 85}
print_values(student)
```

```python
def count_element(t, element):
    count = 0
    for i in t:
        if i == element:
            count = count + 1
    return count

print("Count is:", count_element((1, 2, 3, 2, 2, 4), 2))
```

```python
def reverse_list(lst):
    new_list = []
    for i in range(len(lst) - 1, -1, -1):
        new_list.append(lst[i])
    return new_list

print("Reversed list:", reverse_list([1, 2, 3, 4]))
```
