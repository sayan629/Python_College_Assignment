```python
numbers=[1,2,3,4,5]
even=[]
odd=[]
for n in numbers:
    if n%2==0:
        even.append(n)
    else:
        odd.append(n)
print("Even: ",even)
print("Odd: ",odd)
```

```python
numbers = [1, 2, 2, 3, 1, 4]
count = {}
for n in numbers:
    if n in count:
        count[n] = count[n] + 1
    else:
        count[n] = 1

print(count)
```

```python
marks = {
    188: 35,
    189: 70,
    190: 40
}
passed = 0

for roll in marks:
    if marks[roll] >= 40:
        print("Roll", roll, "Pass")
        passed = passed + 1
    else:
        print("Roll", roll, "Fail")

print("Total Passed:", passed)
```

```python
mylist = []
while True:
    print("1.Add  2.Remove  3.Show  4.Exit")
    choice = int(input("Enter choice: "))
    if choice == 1:
        item = input("Enter item: ")
        mylist.append(item)
    elif choice == 2:
        item = input("Enter item to remove: ")
        if item in mylist:
            mylist.remove(item)
    elif choice == 3:
        print(mylist)
    elif choice == 4:
        break
```

```python
##### inventory = {}

while True:
    print("\n1. Add Item")
    print("2. Update Quantity")
    print("3. Delete Item")
    print("4. Display Low Stock Items")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    # Add Item
    if choice == 1:
        item = input("Enter item name: ")
        qty = int(input("Enter quantity: "))
        inventory[item] = qty
        print("Item added successfully!")

    # Update Quantity
    elif choice == 2:
        item = input("Enter item name: ")
        if item in inventory:
            qty = int(input("Enter new quantity: "))
            inventory[item] = qty
            print("Quantity updated!")
        else:
            print("Item not found!")

    # Delete Item
    elif choice == 3:
        item = input("Enter item name to delete: ")
        if item in inventory:
            del inventory[item]
            print("Item deleted!")
        else:
            print("Item not found!")

    # Display Low Stock
    elif choice == 4:
        found = False
        print("Low Stock Items (quantity < 5):")
        for item in inventory:
            if inventory[item] < 5:
                print(item, ":", inventory[item])
                found = True
        
        if found == False:
            print("No low stock items found.")

    # Exit
    elif choice == 5:
        print("Exiting program...")
        break

    else:
        print("Invalid choice!")
```

```python
numbers = [2, 4, 5, 6, 7]
prime = []
nonprime = []
for n in numbers:
    if n < 2:
        nonprime.append(n)
    else:
        isprime = True
        for i in range(2, n):
            if n % i == 0:
                isprime = False
        if isprime:
            prime.append(n)
        else:
            nonprime.append(n)

print("Prime:", prime)
print("Non Prime:", nonprime)
```

```python
matrix = [[1, 2], [3, 4]]
total = 0
maximum = matrix[0][0]

for row in matrix:
    for num in row:
        total = total + num
        if num > maximum:
            maximum = num

print("Sum:", total)
print("Maximum:", maximum)
```
