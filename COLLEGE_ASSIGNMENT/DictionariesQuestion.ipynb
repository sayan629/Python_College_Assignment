```python
# 1.Create a dictionary of 5 students with their marks and display it.
#     a) Access and print the value of a given key from a dictionary. 
#     b) Add a new key-value pair to an existing dictionary.
#     c) Update the value of an existing key. 
#     d) Delete a key from a dictionary using del. 
#     e) Check whether a key exists in a dictionary. 
#     f) Print all keys and values separately. 
#     g) Find the length of a dictionary. 
#     h) Print all key-value pairs using a for loop.
#     i) Find the student with highest marks.
# 2.Count frequency of each character in a string using dictionary.
# 3.Create a dictionary of numbers and their squares (1 to 10).
# 4.Create nested dictionary for student details (name, marks, age).
# 5.Convert two lists into a dictionary. Example:
# keys = ['a','b','c']
# values = [1,2,3]
```

```python
students = {
    "Sayan": 85,
    "Ruhon": 90,
    "SOham": 78,
    "Shlok": 88,
    "Priya": 92,
    "Shreya": 80
}
print("Students Dictionary:", students)

print("Marks of Priya:", students["Priya"])

students["Aman"] = 75
print("After adding Aman:", students)

students["Sayan"] = 89
print("After updating Sayan:", students)

del students["SOham"]
print("After deleting SOham:", students)

if "Shlok" in students:
    print("Shlok exists")
else:
    print("Shlok does not exist")

print("All Keys:", students.keys())
print("All Values:", students.values())

print("Total students:", len(students))

for name in students:
    print(name, ":", students[name])

highest = max(students, key=students.get)
print("Highest marks student:", highest)
print("Marks:", students[highest])
```

```python
name=input("Enter a String")
frequency={}
for ch in name:
    if ch in frequency:
        frequency[ch]+=1
    else:
        frequency[ch]=1
print("\n Character Frequencies:")
for key, value in frequency.items():
    print(key,":",value)
```

```python
squares={}
for i in range(1,11):
    squares[i]=i*i
print("Numbers and their squares:")
print(squares)
```

```python
students = {
    "student1": {
        "name": "Sayan",
        "marks": 88,
        "age": 21
    },
    "student2": {
        "name": "Shlok",
        "marks": 92,
        "age": 20
    },
    "student3": {
        "name": "Ruhon",
        "marks": 85,
        "age": 22
    }
}
print("Student Details:")
for key, value in students.items():
    print(key, ":", value)

print("\nMarks of Shlok:", students["student2"]["marks"])
```
